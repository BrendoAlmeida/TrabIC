# Execução da primeira entrega

Este notebook chama o código Python em `src/instancias`. Ele importa o modelo de referência e gera os dois arquivos da atividade em `data/instancias/`.

```python
from pathlib import Path
import sys

raiz = Path.cwd().resolve()
if not (raiz / "src").is_dir():
    raiz = raiz.parent
sys.path.insert(0, str(raiz / "src"))

from instancias import carregar_instancia, gerar_instancias
```

## Ler o exemplo

O importador lê o formato original do modelo, sem tipo de prova predefinido para as escolas.

```python
modelo = carregar_instancia(raiz / "exemplos" / "Modelo.txt")
print(f"Modelo importado: {len(modelo.candidatos)} candidatos e {len(modelo.escolas)} escolas.")
```

## Gerar e carregar as instâncias

A semente fixa torna os arquivos reproduzíveis. Para criar outro conjunto, altere `semente`.

```python
arquivos = gerar_instancias(raiz / "data" / "instancias", semente=2026)

for arquivo in arquivos:
    instancia = carregar_instancia(arquivo)
    capacidade_total = sum(escola.capacidade for escola in instancia.escolas)
    print(
        f"{arquivo.name}: {len(instancia.candidatos)} candidatos, "
        f"{len(instancia.escolas)} escolas, {capacidade_total} lugares."
    )
```
