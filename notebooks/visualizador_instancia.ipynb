# Visualizador de instância

O gráfico permite aproximar e afastar a visualização, ocultar tipos pela legenda e passar o cursor sobre cada ponto para consultar seus dados. Escolas não têm tipo de prova predefinido na instância.

```python
from pathlib import Path
import sys

raiz = Path.cwd().resolve()
if not (raiz / "src").is_dir():
    raiz = raiz.parent
sys.path.insert(0, str(raiz / "src"))

from instancias import carregar_instancia
from instancias.visualizacao import grafico_candidatos_por_tipo, grafico_instancia
```

## Escolher o arquivo

Altere `nome_instancia` para `instancia_1.txt` ou `instancia_2.txt` e execute as células seguintes.

```python
nome_instancia = "instancia_2.txt"
caminho_instancia = raiz / "data" / "instancias" / nome_instancia
instancia = carregar_instancia(caminho_instancia)
capacidade_total = sum(escola.capacidade for escola in instancia.escolas)
print(f"{nome_instancia}: {len(instancia.candidatos)} candidatos, {len(instancia.escolas)} escolas")
print(f"Capacidade total: {capacidade_total}; ocupação: {len(instancia.candidatos) / capacidade_total:.1%}")
```

```python
grafico_instancia(instancia, titulo=nome_instancia).show()
```
